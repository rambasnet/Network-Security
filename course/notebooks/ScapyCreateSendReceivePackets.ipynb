## Send and Receive Packets
- Send and receive functions are the heart of Scapy
- sr() - send L3 packets and receive answers; returns answered and unanswered packets
- sr1() - returns only one packet that answered the sent packet
- sr() and sr1() functions can only send layer 3 L3 packets (IP, HTTP, etc.)
- srp() - send and receive L2 packets (ARP, Ehternet, 802.3, etc)
- srp1() - send and receive one layer 2 packet

````python
response = sr(httpPkt)
````

````python
# get the answer tuple
ans = response[0]
````

````python
help(ans)
````

````python
ans.show()
````

````python
print(ans[0])
````

````python
ans[0]
````

````python
# if you want to get the complete http response data and parse it
# use HTTP-Client class

from scapy.layers.http import HTTP_Client, HTTPResponse

client = HTTP_Client()

try:
    response = client.request("http://192.168.100.200:5555")

    if response is not None and response.haslayer(HTTPResponse):
        http = response[HTTPResponse]

        print("Status:", http.Status_Code.decode())
        print("Content-Type:", http.Content_Type)

        body = bytes(http.payload)
        print(body.decode("utf-8", errors="replace"))
finally:
    client.close()
````

## TCP/IP full three-way handshake

- Scapy crafts packets outside the kernel’s TCP connection tracking
- When the SYN-ACK arrives, the kernel can send a RST because it does not recognize the connection.
- Scapy’s documentation describes this issue. [https://scapy.readthedocs.io/en/latest/troubleshooting.html?utm_source=chatgpt.com](https://scapy.readthedocs.io/en/latest/troubleshooting.html?utm_source=chatgpt.com)

````python
! iptables -I OUTPUT -p tcp -d 192.168.100.200 \
  --sport 40000 --dport 5555 --tcp-flags RST RST -j DROP
````

### send SYN packet

````python
# Create syn packet
server_host = '192.168.100.200'
server_port = 5555
client_port = 40000
client_isn = 1000 # initial sequence number
ip = IP(dst=server_host)
syn = ip/TCP(sport=client_port, dport=server_port, flags='S', seq=client_isn)
````

````python
syn.show()
````

````python
# Receive SYN-ACK packet
syn_ack = sr1(syn, timeout=3)
````

````python
syn_ack.show()
````

````python
# validate reponse before continuing
assert syn_ack is not None, "No response received"
assert syn_ack.haslayer(TCP), "Response is not TCP"
assert not (int(syn_ack[TCP].flags) & 0x04), "Server sent RST"
assert (int(syn_ack[TCP].flags) & 0x12) == 0x12, "Expected SYN-ACK"
assert syn_ack[TCP].ack == client_isn + 1, "Unexpected acknowledgment"
````

````python
# see key fields
print("Flags:", syn_ack[TCP].flags)
print("Server sequence:", syn_ack[TCP].seq)
print("Server acknowledgment:", syn_ack[TCP].ack)
````

### Create and send ACK packet to complete the 3-way handshake

````python
# send the ACK
ack = ip / TCP(
    sport=client_port,
    dport=server_port,
    flags="A",
    seq=syn_ack[TCP].ack,
    ack=(syn_ack[TCP].seq + 1) % (2**32),
)
````

### Before sending SYN, run this in the Client Terminal

```bash
sudo tcpdump -i any -nn -S 'host 192.168.100.200 and tcp port 5555'
```

### On the server, run one of the following commands to see the established connection

```bash
ss -tn state established sport = :5555
netstat -na
```
- note you may see SYN-RECV --TCP half-open connection--, if the ACK packet is not sent

### Close the established connection to re-initiate the 3-way handshake

```bash
sudo ss -K -t state established '( sport = :5555 and dport = :40000 )'
```
- if the above command doesn't close the connection and clear the cache, try shutting down and running the server container

````python
# just send because pure ACK normally does NOT receive an immediate response
send(ack, verbose=False)
````

## Create ping request
